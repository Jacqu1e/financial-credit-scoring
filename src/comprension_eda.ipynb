```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo gráfico
plt.style.use('seaborn-v0_8-whitegrid')
plt.rcParams['figure.figsize'] = (10, 5)

# Cargar el dataset usando el encoding que validamos previamente
df = pd.read_csv('../Base_de_datos.csv', encoding='latin1')

# Dimensiones e inspección de estructura y nulos
print(f"Dimensiones: {df.shape[0]} filas | {df.shape[1]} columnas\n")

resumen_calidad = pd.DataFrame({
    'Tipo': df.dtypes,
    'Valores Nulos': df.isnull().sum(),
    '% Nulos': (df.isnull().sum() / len(df)) * 100
})
resumen_calidad
```

```python
# Variables numéricas y categóricas
cols_num = df.select_dtypes(include=[np.number]).columns.tolist()
cols_cat = df.select_dtypes(include=['object', 'category']).columns.tolist()

# Histogramas para variables numéricas
df[cols_num].hist(bins=20, figsize=(14, 8), color='#34495e', edgecolor='white')
plt.suptitle('Análisis Univariable: Distribución de variables numéricas', fontsize=14)
plt.tight_layout()
plt.show()
```

```python
# Matriz de Correlación
plt.figure(figsize=(10, 6))
sns.heatmap(df[cols_num].corr(), annot=True, fmt='.2f', cmap='coolwarm', linewidths=0.5)
plt.title('Análisis Multivariable: Matriz de Correlación de Pearson')
plt.show()
```
